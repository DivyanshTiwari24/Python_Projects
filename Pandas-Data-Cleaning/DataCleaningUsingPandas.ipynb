```python
import pandas as pd
```

```python
df = pd.read_excel(r'C:\Users\divya\OneDrive\Desktop\Project\Customer Call List.xlsx')
```

```python
df.head()
```

```python
df_new = df['Phone_Number'].str.replace(r'\D' , '',regex = True)
```

```python
df['Phone_Number'] = df['Phone_Number'].where(df_new.isna(),df_new)
```

```python
df['Phone_Number'] = df['Phone_Number'].astype(str)
```

```python
df['Phone_Number'] = df['Phone_Number'].str[:3] + '-' + df['Phone_Number'].str[3:6]  + '-' + df['Phone_Number'].str[6:10]  
```

```python
df.head()
```

```python
df.groupby('Do_Not_Contact')['CustomerID'].count()
```

```python
df['Paying Customer'] = df['Paying Customer'].apply(lambda x: 'Yes' if str(x).startswith(('Y','y')) else 'No' if str(x).startswith(('N','n')) else x) 
```

```python
df['Do_Not_Contact'] = df['Do_Not_Contact'].apply(lambda x: 'Yes' if str(x).startswith(('Y','y')) else 'No' if str(x).startswith(('N','n')) else x) 
```

```python
df.tail()
```

```python
df[['Street','State']] = df['Address'].str.split(',' , n=1,expand = True)
```

```python
df = df.drop_duplicates()
```

```python
df.drop(columns = {'Not_Useful_Column'},inplace = True)
```

```python
 df.head()
```

```python
df['Last_Name'] = df['Last_Name'].str.replace(r'^[^a-zA-Z]+', '', regex=True)
```

```python
df['State'] = df['State'].fillna('Unknown')
```

```python
df['State'] = df['State'].fillna('Not Mentioned')
```

```python
df['Phone_Number'].dtype
```

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('--','',regex=True)
```

```python
df['Phone_Number']=df['Phone_Number'].fillna('')
```

```python
df = df[df['Phone_Number'] != '']
```
